```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("criminalita_densita_regioni.csv")

correlazione = df["densita_ab_kmq"].corr(df["operazioni_per_100k_ab"])

fig = px.scatter(
    df,
    x="densita_ab_kmq",
    y="operazioni_per_100k_ab",
    text="regione",
)

fig.update_traces(
    marker=dict(size=9, color="#8c3b45"),
    textposition="top center",
    textfont=dict(size=9),
)

# rinomina solo alcune etichette mostrate (i dati restano invariati)
etichette = df["regione"].replace({
    "Trentino-Alto Adige": "Trentino",
    "Friuli Venezia Giulia": "Friuli",
    "Emilia Romagna": "Emilia",
})
fig.update_traces(text=etichette)

# posizioni personalizzate solo per queste 3 regioni, le altre restano "top center"
posizioni_personalizzate = {
    "Calabria": "middle right",
    "Molise": "bottom center",
    "Trentino-Alto Adige": "bottom center",
}
posizioni = [posizioni_personalizzate.get(regione, "top center") for regione in df["regione"]]
fig.update_traces(textposition=posizioni)

# range asse X compresso attorno ai dati reali (invece del padding di default)
x_min, x_max = df["densita_ab_kmq"].min(), df["densita_ab_kmq"].max()
margine = (x_max - x_min) * 0.08

fig.update_layout(
    xaxis_title="Densità abitativa (ab./km²)",
    yaxis_title="Operazioni antidroga ogni 100.000 abitanti",
    xaxis=dict(range=[x_min - margine, x_max + margine], dtick=50),
    template="simple_white",
    width=700,   # grafico più stretto: riduce la distanza fisica tra i valori sull'asse X
    height=600,
)

# 3. NOTA FONTE (in basso)
fig.add_annotation(
    text=(f""),
    xref="paper", yref="paper",
    x=0, y=-0.18,
    showarrow=False,
    font=dict(size=10, color="#888888"),
    align="left",
)

fig.show()
```
