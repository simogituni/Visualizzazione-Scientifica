```python
import pandas as pd
from plotly.subplots import make_subplots
import plotly.graph_objects as go

df_raw = pd.read_csv("sostanze.csv")

gruppi = {
    "Cannabinoidi": ["Cannabinoids"],
    "Catinoni": ["Cathinones"],
    "Oppioidi": ["Opioids"],
    "Benzodiazepine": ["Benzodiazepines"],
    "Stimolanti minori": ["Phenethylamines", "Piperazines", "Aminoindanes", "Arylalkylamines"],
    "Allucinogeni e altro": ["Tryptamines", "Arylcyclohexylamines", "Plants and extracts", "Other substances"],
}

df = pd.DataFrame({"anno": df_raw["Year"]})
for nome_gruppo, colonne in gruppi.items():
    df[nome_gruppo] = df_raw[colonne].sum(axis=1)

categorie = list(gruppi.keys())
y_max = df[categorie].values.max() * 1.15  # stessa scala Y per tutti i pannelli

colori = {
    "Cannabinoidi": "#5d8a66",          # verde desaturato
    "Catinoni": "#8c3b45",              # bordeaux desaturato
    "Oppioidi": "#5d7ea8",              # blu desaturato
    "Benzodiazepine": "#AA69A4",        # viola desaturato
    "Stimolanti minori": "#c99a4a",     # ocra desaturato
    "Allucinogeni e altro": "#7a8a8a",  # grigio-verde desaturato
}


fig = make_subplots(
    rows=2, cols=3,
    subplot_titles=categorie,
    shared_yaxes=False,  # ogni pannello mostra i propri numeri sull'asse Y
)

posizioni = [(1, 1), (1, 2), (1, 3), (2, 1), (2, 2), (2, 3)]

for categoria, (riga, colonna) in zip(categorie, posizioni):
    fig.add_trace(
        go.Bar(
            x=df["anno"], y=df[categoria],
            marker_color=colori[categoria],
            showlegend=False,
        ),
        row=riga, col=colonna,
    )

fig.update_yaxes(range=[0, y_max], showticklabels=True, tick0=0, dtick=10)
fig.update_xaxes(
    tickvals=[2005, 2010, 2015, 2020, 2023],
    type="category",
    showticklabels=True,
)

fig.update_layout(
    template="simple_white",
    height=500,
    margin=dict(b=90),
)

fig.add_annotation(
    text=("Fonte: EUDA - European Drug Report 2024 "
          "\"Stimolanti minori\" = fenetilamine, piperazine, aminoindani, arilalchilamine. "
          "\"Allucinogeni e altro\" = triptamine, arilcicloesilamine, piante/estratti, altre sostanze."),
    xref="paper", yref="paper",
    x=0, y=-0.20,
    showarrow=False,
    font=dict(size=10, color="#888888"),
    align="left",
)

fig.show()
```
