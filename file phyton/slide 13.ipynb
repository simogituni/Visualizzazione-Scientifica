```python
import pandas as pd
import plotly.graph_objects as go

ALTEZZA_GRAFICO = 600

df = pd.read_csv("decessi_mondo_per_eta.csv")

# limitiamo la serie al 1990 in poi, come da titolo della slide
df = df[df["anno"] >= 1990]

# ordine di impilamento: dalla fascia più giovane (in basso) alla più anziana
# (in cima), come nel grafico originale di Our World in Data
fasce = ["<20", "20-29", "30-49", "50-69", ">=70"]

color_map = {
    "<20": "#f2c9b0",
    "20-29": "#e0a180",
    "30-49": "#c97c5d",
    "50-69": "#a8544a",
    ">=70": "#7a3235",
}

fig = go.Figure()

for fascia in fasce:
    dati_fascia = df[df["fascia_eta"] == fascia].sort_values("anno")
    fig.add_trace(go.Scatter(
        x=dati_fascia["anno"],
        y=dati_fascia["decessi_stimati"],
        name=fascia,
        mode="lines",
        stackgroup="uno",  # attiva l'impilamento (area chart)
        line=dict(width=0.5, color=color_map[fascia]),
        fillcolor=color_map[fascia],
    ))

fig.update_layout(
    height=ALTEZZA_GRAFICO,  # altezza del grafico
    xaxis_title="Anno",
    yaxis_title="N. decessi stimati",
    xaxis=dict(dtick=5),
    
    yaxis=dict(side="left", tickformat="~s"),
    # asse Y gemello a destra: "matches='y'" lo tiene sincronizzato in scala
    # e tick con l'asse di sinistra, quindi mostra sempre gli stessi valori,
    # rendendo più facile leggere il grafico senza dover guardare solo a sinistra
    yaxis2=dict(
        overlaying="y",
        side="right",
        matches="y",
        showgrid=False,
        showticklabels=True,
        tickformat="~s",
        title="N. decessi stimati",
    ),
    template="simple_white",
    legend_title_text="Fascia d'età",
)

fig.add_annotation(
    text=(""),
    xref="paper", yref="paper",
    x=0, y=-0.22,
    showarrow=False,
    font=dict(size=10, color="#888888"),
    align="left",
)

fig.show()
```
